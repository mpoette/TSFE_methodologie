```python
with pl.Config(tbl_cols = -1, tbl_rows = -1):
    df_full = pl.read_parquet("Datasets/df_with_calculated_features.parquet")
    print(df_full.head(1))

    df_partial = pl.read_parquet("Datasets/temporal_tailored_imputation.parquet")
    print(df_partial.head(1))
```

Là on passe donc pour les données amputées de 65 features à 50 puisque les 15 derniers ne sont pas présents sur le dataframe initial.

```python
df_mask = pl.DataFrame({
    col : (df_full[col].is_not_null() & df_partial[col].is_not_null()).cast(pl.Int8)
    for col in df_full.columns
})
```
